```python
import pandas as pd
from sylphy.embedding_extractor import create_embedding
from sklearn.preprocessing import normalize
```

```python
df_data = pd.read_csv("../pipelines/data/sequences.csv")
df_data.head()
```

```python
embedder = create_embedding(
    model_name="facebook/esm2_t6_8M_UR50D",
    dataset=df_data,
    column_seq="sequence",
    name_device="cuda",
    precision="fp32"
)
```

```python
embedder.run_process(batch_size=16, pool="mean") 
```

```python
df_encoders = embedder.coded_dataset
df_encoders
```

```python
X_used = normalize(df_encoders.drop(columns=["sequence"]).values, norm="l2")
X_used
```
