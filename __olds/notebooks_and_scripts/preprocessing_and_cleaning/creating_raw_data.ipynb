#### Antioxidant protein dataset formatting and identifier generation

This script formats the filtered antioxidant protein dataset, generates unique identifiers for each sequence, and prepares the data for downstream pipeline usage.

- Overview
    - Task: Antioxidant protein dataset formatting
    - Input: Filtered dataset (CSV) containing sequences and labels
    - Output: Formatted dataset (CSV) with sequence identifiers
- Process:
    - Load filtered dataset
    - Select relevant columns (sequence and label)
    - Generate unique identifiers for each sequence
    - Append identifier column to dataset
    - Export formatted dataset

- Auxiliary variables

```python
path_data = "../../processed_dataset"
path_export = "../../pipelines/data"
```

- Read data

```python
df_data_filtered = pd.read_csv(f"{path_data}/processed_data/processed_dataset.csv")
df_data_filtered
```

- Create identifier

```python
df_data_filtered = df_data_filtered[["sequence", "label"]]
df_data_filtered["id"] = [f"seq_{i+1}" for i in range(df_data_filtered.shape[0])]
df_data_filtered.head()
```

- Export data

```python
df_data_filtered.to_csv(f"{path_export}/sequences.csv", index=False)
```
