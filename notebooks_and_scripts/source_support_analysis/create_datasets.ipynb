#### Dataset Subset Construction Based on Source Support

This notebook generates sequence-level dataset subsets according to the number of valid sources supporting each record. The resulting datasets include single-source, multi-source, high-support, and full-consensus subsets, which are exported as standardized input files for downstream machine learning pipelines.

- Auxiliary variables and functions

```python
path_data = "../../processed_dataset"
path_export = "../../pipelines/data"
```

```python
def prepare_dataset(df, prefix="seq"):
    df_out = df[["sequence", "label"]].copy()
    df_out["id"] = [f"seq_{prefix}_{i+1}" for i in range(df_out.shape[0])]
    return df_out
```

- Read data

```python
df_data_filtered = pd.read_csv(f"{path_data}/processed_data/processed_dataset.csv")
df_data_filtered
```

- Construct subsets of the dataset based on source support

```python
single_source = df_data_filtered[df_data_filtered["n_valid_sources"] == 1]
multi_source = df_data_filtered[df_data_filtered["n_valid_sources"] >= 2]
high_support = df_data_filtered[df_data_filtered["n_valid_sources"] >= 3]
full_consensus = df_data_filtered.copy()
```

```python
datasets = {
    "single_source": single_source,
    "multi_source": multi_source,
    "high_support": high_support,
    "full_consensus": full_consensus
}

datasets_ready = {}

for name, df in datasets.items():
    datasets_ready[name] = prepare_dataset(df, prefix=name)
```

```python
for name, df in datasets_ready.items():
    df.to_csv(f"{path_export}/sequences_{name}.csv", index=False)
```
