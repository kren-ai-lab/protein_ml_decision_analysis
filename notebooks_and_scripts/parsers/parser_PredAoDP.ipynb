```python
import warnings
import pandas as pd
warnings.filterwarnings("ignore")
from building_models.commons_functions.parsers_commons import ParsersCommons
from building_models.utils.constants import COLUMNS_TO_WORK
from building_models.utils.utils_functions import UtilsFunctions
```

#### Antioxidant protein dataset preprocessing and metadata generation

This script processes antioxidant protein data from multiple files within a single source, assigns labels using different rules depending on the file, handles duplicated sequences, and generates a structured dataset along with metadata.

- Overview
    - Task: Antioxidant protein classification dataset preparation
    - Source: PredAoDP dataset
    - Input: Multiple FASTA-like files containing protein sequences and metadata Excel file
    - Output: Processed dataset (CSV) and metadata file (JSON)
- Process:
    - Read protein sequences from multiple input files
    - Assign labels using different strategies (fixed labels, identifier-based rules, and index-based splits)
    - Standardize data by removing identifier columns
    - Concatenate all datasets into a single dataset
    - Check for duplicated sequences and label consistency
    - Concatenate consistent duplicates with unique sequences
    - Load and organize source metadata
    - Export processed data and metadata

- Auxiliary variables and functions

```python
def add_label_based_on_id(id_value):
    if "non-antioxidant" in id_value:
        return 0
    else:
        return 1
```

```python
path_export = "../../processed_dataset"
path_input = "../../raw_dataset"
metadata_file = "../../raw_dataset/raw_data_description.xlsx"
name_task = "antioxidant_classification"
name_source = "PredAoDP"
```

- Read docs

```python
df_anti = ParsersCommons.read_fasta_doc(f"{path_input}/{name_source}/anti.txt")
df_anti["label"] = 1
df_anti = df_anti.drop(columns=["id"])
df_anti.head()
```

```python
df_Antioxident666 = ParsersCommons.read_fasta_doc(f"{path_input}/{name_source}/Antioxident666.txt")
df_Antioxident666["label"] = df_Antioxident666["id"].apply(add_label_based_on_id)
df_Antioxident666 = df_Antioxident666.drop(columns=["id"])
df_Antioxident666.head()
```

```python
df_independent_dataset = ParsersCommons.read_fasta_doc(f"{path_input}/{name_source}/independent dataset.txt")
print(df_independent_dataset.shape)
df_independent_dataset_neg = df_independent_dataset[:392]
df_independent_dataset_pos = df_independent_dataset[392:]
df_independent_dataset_neg["label"] = 0
df_independent_dataset_pos["label"] = 1

df_independent_dataset = pd.concat([df_independent_dataset_neg, df_independent_dataset_pos], axis=0, ignore_index=True)
df_independent_dataset = df_independent_dataset.drop(columns=["id"])
df_independent_dataset.head()
```

```python
df_independent_dataset_2 = ParsersCommons.read_fasta_doc(f"{path_input}/{name_source}/Indpendent dataset.txt")
df_independent_dataset_2.head()
```

```python
df_independent_dataset_2_anti = df_independent_dataset_2[df_independent_dataset_2["id"].str.startswith("anti")]
df_independent_dataset_2_nonanti = df_independent_dataset_2[~df_independent_dataset_2["id"].str.startswith("anti")]
```

```python
df_independent_dataset_2_anti["label"] = 1
df_independent_dataset_2_nonanti["label"] = 0

df_independent_dataset_2 = pd.concat([df_independent_dataset_2_anti, df_independent_dataset_2_nonanti], 
                                     axis=0, ignore_index=True)
df_independent_dataset_2 = df_independent_dataset_2.drop(columns=["id"])

df_independent_dataset_2.head()
```

```python
df_nonanti = ParsersCommons.read_fasta_doc(f"{path_input}/{name_source}/nonanti.txt")
df_nonanti["label"] = 0
df_nonanti = df_nonanti.drop(columns=["id"])

df_nonanti.head()
```

- Concat data and check duplicates

```python
df_concat = pd.concat([
    df_nonanti, 
    df_independent_dataset_2, 
    df_independent_dataset,
    df_anti,
    df_Antioxident666], ignore_index=True, axis=0)
df_concat.shape, df_concat["sequence"].unique().shape
```

```python
df_consistent_duplicates, df_errors, df_unique = ParsersCommons.processing_duplicated(
    df_concat, group_seq= "sequence",
    label_col= "label")
df_consistent_duplicates.shape, df_errors.shape, df_unique.shape
```

```python
data_correct = pd.concat([df_consistent_duplicates, df_unique], axis=0, ignore_index=True)
data_correct = data_correct.drop(columns=["n_duplicates"])
data_correct.head()
```

- Reading metadata

```python
metadata_file = ParsersCommons.read_metadata(metadata_file, name_source=name_source, columns_to_select=COLUMNS_TO_WORK)
metadata_file.head()
```

```python
dict_metadata = ParsersCommons.create_metadata_from_file(metadata_file)
dict_metadata
```

```python
dict_metadata['number_of_records'] = df_concat.shape[0]
dict_metadata['number_of_collected_sequences'] = df_concat.shape[0]
dict_metadata['number_of_unique_sequences'] = data_correct.shape[0]
dict_metadata['positive_examples'] = data_correct[data_correct["label"] == 1].shape[0]
dict_metadata['negative_examples'] = data_correct[data_correct["label"] == 0].shape[0]
dict_metadata['number_of_sequences_with_errors'] = df_errors.shape[0]
dict_metadata
```

- Export data

```python
UtilsFunctions.make_directory(f"{path_export}/{name_task}/{name_source}")
UtilsFunctions.export_json(f"{path_export}/{name_task}/{name_source}/metadata_{name_source}.json", dict_metadata)
data_correct.to_csv(f"{path_export}/{name_task}/{name_source}/processed_data.csv", index=False)
```
