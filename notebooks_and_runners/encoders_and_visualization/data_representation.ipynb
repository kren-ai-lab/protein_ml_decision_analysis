```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import pandas as pd
from building_models.numerical_representation.encoders import EncodeSequences
from building_models.utils.utils_functions import UtilsFunctions
```

```python
path_data = "../../processed_dataset/"
```

```python
df_data = pd.read_csv(f"{path_data}processed_data/processed_dataset.csv")
df_data = df_data[["sequence", "label"]]
df_data.head(5)
```

```python
df_errors = pd.read_csv(f"{path_data}processed_data/sequences_with_errors.csv")
df_errors = df_errors[["sequence"]]
df_errors.head(5)
```

```python
df_unlabeled = pd.read_csv(f"{path_data}processed_data/unlabeled_data.csv")
df_unlabeled = df_unlabeled[["sequence"]]
df_unlabeled.head(5)
```

- Getting baseline representation

```python
UtilsFunctions.make_directory(f"{path_data}encoded_data/one_hot")
```

```python
for data in [(df_data, "full_data"), (df_errors, "error_data"), (df_unlabeled, "unlabeled_data")]:
    dataset = data[0]
    name_export = data[1]

    df_one_hot = EncodeSequences.encode_dataset(df=dataset, column_with_sequence="sequence", method="one_hot")
    df_one_hot = df_one_hot.merge(right=df_data, on="sequence")
    df_one_hot = df_one_hot.drop(columns=["is_canon"])
    df_one_hot.to_csv(f"{path_data}encoded_data/one_hot/{name_export}.csv", index=False)
```

```python
UtilsFunctions.make_directory(f"{path_data}encoded_data/esm2_t6_8M_UR50D")
```

```python
for data in [(df_data, "full_data"), (df_errors, "error_data"), (df_unlabeled, "unlabeled_data")]:
    dataset = data[0]
    name_export = data[1]

    df_embedding = EncodeSequences.get_embeddings(df=dataset, column_with_sequence="sequence")
    df_embedding = df_embedding.merge(right=df_data, on="sequence")
    df_embedding = df_embedding.drop(columns=["is_canon"])
    df_embedding.to_csv(f"{path_data}encoded_data/esm2_t6_8M_UR50D/{name_export}.csv", index=False)
```
